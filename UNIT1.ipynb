```python
#PROGRAM ON MATCH
num=int(input("enter the number"))
match num:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
```

```python
#PRINT 1 TO 10 NUMBER USING WHILE LOOP
i=1
while(i<=10):
  print(i)
  i+=1
```

```python
#PRINT 10 TO 1 NUMBERS USING WHILE
i=10
while(i>0):
  print(i)
  i-=1
```

```python
#PRINT EVEN NUMBERS FROM 1 TO 10
i=2
while(i<=10):
  print(i)
  i+=2
```

```python
#PRINT REVERSE EVEN NUMBERS FROM 1 TO 10
i=10
while(i>0):
  print(i)
  i-=2
```
