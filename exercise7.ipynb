```python
#methods on set
set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5}

print("Original set1", set1)

set1.add(4)
print("After add()", set1)

set1.update([5, 6])
print("After update()", set1)

set1.remove(6)
print("After remove()", set1)

set1.discard(10)
print("After discard()", set1)

x = set1.pop()
print("Popped element", x)
print("After pop()", set1)

print("Is set1 a subset of set2", set1.issubset(set2))

print("Is set2 a superset of set1", set2.issuperset(set1))

set3 = {10, 20, 30}
print("Is set1 disjoint with set", set1.isdisjoint(set3))
```
