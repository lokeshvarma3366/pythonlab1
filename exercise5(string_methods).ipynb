#string methods


```python
a="hello world"
b=a.split()
print(b)
```

```python
c="".join(b)
print(c)
```

```python
d=a.strip()
print(d)
```

```python
e=a.upper()
print(e)
```

```python
f=a.lower()
print(f)
```

```python
g=a.replace("hello","bye")
print(g)
```

```python
h=a.find('1')
print(h)
```

```python
i=a.count('4')
print(i)
```
